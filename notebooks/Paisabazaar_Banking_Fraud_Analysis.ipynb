#### Chart - 1 : Count Plot - Distribution of Credit Score

````python
# Chart 1 - Count plot
plt.figure(figsize=(8,5))

sns.countplot(
    data=df_clean,
    x='Credit_Score',
    order=df_clean['Credit_Score'].value_counts().index,
    palette='Set2'
)

plt.title('Distribution of Credit Score', fontsize=14)
plt.xlabel('Credit Score')
plt.ylabel('Number of Customers')

plt.show()
````

````python
# For data specific insights
df_clean['Credit_Score'].value_counts()
````

##### 1. Why did you pick the specific chart?

A count plot was chosen because Credit_Score is the target variable and is categorical in nature, consisting of three classes: Poor, Standard, and Good. This provides a clear overview of the target variable before exploring the factors that influence it.

##### 2. What is/are the insight(s) found from the chart?

The chart shows that the majority of customers belong to the Standard credit score category, accounting for 53.17% of the total customer base (53,174 customers). Customers with a Poor credit score represent 29.00% (28,998 customers), while only 17.83% (17,828 customers) have a Good credit score. This indicates that more than half of the customers maintain an average credit profile, whereas a relatively smaller proportion demonstrates excellent creditworthiness.
